```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Airport": [
        "Kuala Lumpur",
        "Singapore",
        "Bangkok",
        "Jakarta",
        "Tokyo",
        "Seoul",
        "Hong Kong",
        "Dubai",
        "Doha",
        "Sydney"
    ],

    "Country": [
        "Malaysia",
        "Singapore",
        "Thailand",
        "Indonesia",
        "Japan",
        "South Korea",
        "Hong Kong",
        "UAE",
        "Qatar",
        "Australia"
    ],

    "Passengers_Millions": [
        62,
        68,
        58,
        55,
        75,
        71,
        61,
        66,
        45,
        42
    ]
}

df = pd.DataFrame(data)

df
```

```python
df.loc[df["Passengers_Millions"].idxmax()]
```

```python
df.sort_values(
    "Passengers_Millions",
    ascending=False
)
```

```python
df["Passengers_Millions"].mean()
```

```python
df_sorted = df.sort_values(
    "Passengers_Millions",
    ascending=True
)

df_sorted.plot(
    x="Airport",
    y="Passengers_Millions",
    kind="barh"
)

plt.title("Airport Passenger Traffic")
plt.xlabel("Passengers (millions)")
plt.ylabel("Airport")

plt.show()
```

```python
data = {
    "Airport": [
        "Kuala Lumpur",
        "Singapore",
        "Bangkok",
        "Jakarta",
        "Tokyo",
        "Seoul",
        "Hong Kong",
        "Dubai",
        "Doha",
        "Sydney"
    ],

    "Country": [
        "Malaysia",
        "Singapore",
        "Thailand",
        "Indonesia",
        "Japan",
        "South Korea",
        "Hong Kong",
        "UAE",
        "Qatar",
        "Australia"
    ],

    "Passengers_Millions": [
        62, 68, 58, 55, 75,
        71, 61, 66, 45, 42
    ],

    "International_Percent": [
        72, 90, 65, 40, 55,
        68, 95, 92, 94, 48
    ],

    "Destinations": [
        140, 160, 130, 110, 180,
        170, 150, 200, 180, 120
    ]
}

df = pd.DataFrame(data)

df
```

```python
df["Passengers_Millions"].corr(
    df["International_Percent"]
)
```
